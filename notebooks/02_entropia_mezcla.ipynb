Calcula la contribución ideal de mezcla para dos componentes y explora límites.

## Función de mezcla

Antes de ejecutar: ¿en qué composición esperas mayor magnitud de ΔS_mix?

```python
import numpy as np
R=8.314
def ds_mix(x1, n=1.0):
    x2=1-x1
    terms=[]
    for x in (x1,x2):
        terms.append(0 if x==0 else x*np.log(x))
    return -n*R*sum(terms)
for x in [0.01,0.1,0.25,0.5,0.75,0.9,0.99]:
    print(x, f"{ds_mix(x):.3f} J/K")
```

## Gráfica

Describe con palabras la simetría y los límites antes de verla.

```python
import numpy as np
import matplotlib.pyplot as plt
x=np.linspace(0.001,0.999,400)
y=np.array([ds_mix(v) for v in x])
plt.figure()
plt.plot(x,y)
plt.xlabel("fracción molar x1")
plt.ylabel("ΔS_mix (J/K por mol total)")
plt.title("Mezcla ideal binaria")
plt.show()
```
