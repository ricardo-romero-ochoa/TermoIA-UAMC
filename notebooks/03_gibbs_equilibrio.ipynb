## Relación entre ΔG° y K

Predice qué ocurre con K cuando ΔG° se hace más negativo.

```python
import numpy as np
R,T=8.314,298.15
for dG0_kJ in [-20,-10,0,10,20]:
    K=np.exp(-(dG0_kJ*1000)/(R*T))
    print(f"ΔG°={dG0_kJ:>3} kJ/mol -> K={K:.3g}")
```

## Dirección según Q

Comprueba el signo de ΔG para Q<K, Q=K y Q>K.

```python
import numpy as np
R,T=8.314,298.15
K=10.0
dG0=-R*T*np.log(K)
for Q in [0.1,1,10,100]:
    dG=dG0+R*T*np.log(Q)
    print(f"Q={Q:>5}: ΔG={dG/1000:.3f} kJ/mol")
```
