# Unión 1:1 de ligando: ocupación y Kd

Modelo mínimo θ = [L]/(Kd+[L]). Úsalo para aprender qué sí y qué no se deduce de una curva de equilibrio.

**Regla:** predice la tendencia antes de ejecutar cada bloque. Después explica si el resultado confirma o corrige tu predicción.

## Límites

Predice θ cuando [L] << Kd, [L]=Kd y [L] >> Kd.

```python
def theta(L,Kd):
    return L/(Kd+L)
Kd=1.0
for L in [0.01,0.1,1,10,100]:
    print(f"L/Kd={L/Kd:>6.2f}, θ={theta(L,Kd):.4f}")
```

## Familia de curvas

Antes de ejecutar: ¿hacia dónde se desplaza la curva si Kd disminuye?

```python
import numpy as np
import matplotlib.pyplot as plt
L=np.logspace(-3,3,400)
plt.figure()
for Kd in [0.1,1,10]:
    plt.semilogx(L, theta(L,Kd), label=f"Kd={Kd}")
plt.xlabel("[L] (unidades relativas)")
plt.ylabel("fracción ocupada θ")
plt.ylim(0,1.02)
plt.legend()
plt.title("Modelo de unión 1:1")
plt.show()
```
